## Choose the feature views

The cell above extracts crop features once, then reloads compatible features on later runs. `patch` uses normalized patch features. `combined` blends normalized patch and CLS features before renormalization and PCA; `cls_weight` controls that blend.

Choose **1–5 slices** below. PCA is fitted once per whole 3D crop and encoder, then projected onto each slice. Colors describe feature variation within a panel and are not aligned between encoders. [PCA method](../docs/visualization.md#pca).


```python
from visualization_utils import plot_features

pca_modes = ["patch", "combined"]
cls_weight = 0.5
num_slices = 5  # Choose 1–5 slices, matching the input preview.

for mode in pca_modes:
    plot_features(
        config, arrays, receipt, kind="pca", mode=mode,
        cls_weight=cls_weight, num_slices=num_slices,
    )
    plt.show()
```
